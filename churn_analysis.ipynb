```python
from scipy.stats import ttest_ind
```

```python
retained = df[df['Churned'] == 0]['Cart_Abandonment_Rate'].dropna()

churned = df[df['Churned'] == 1]['Cart_Abandonment_Rate'].dropna()

t_stat, p_value = ttest_ind(
    retained,
    churned,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
print("Retained mean:", retained.mean())
print("Churned mean:", churned.mean())
print("Difference:", churned.mean() - retained.mean())
```

```python
retained = df[df['Churned'] == 0]['Session_Duration_Avg'].dropna()

churned = df[df['Churned'] == 1]['Session_Duration_Avg'].dropna()

t_stat, p_value = ttest_ind(
    retained,
    churned,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

print("Retained mean:", retained.mean())
print("Churned mean:", churned.mean())
print("Difference:", churned.mean() - retained.mean())
```

```python
retained = df[df['Churned'] == 0]['Customer_Service_Calls'].dropna()

churned = df[df['Churned'] == 1]['Customer_Service_Calls'].dropna()

t_stat, p_value = ttest_ind(
    retained,
    churned,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

print("Retained mean:", retained.mean())
print("Churned mean:", churned.mean())
print("Difference:", churned.mean() - retained.mean())
```

```python
results = []

variables = [
    'Cart_Abandonment_Rate',
    'Session_Duration_Avg',
    'Customer_Service_Calls'
]

for col in variables:

    retained = df[df['Churned'] == 0][col].dropna()
    churned = df[df['Churned'] == 1][col].dropna()

    t_stat, p_value = ttest_ind(
        retained,
        churned,
        equal_var=False
    )

    results.append({
        'Metric': col,
        'Retained Mean': retained.mean(),
        'Churned Mean': churned.mean(),
        'Difference': churned.mean() - retained.mean(),
        'T-statistic': t_stat,
        'P-value': p_value
    })

results_df = pd.DataFrame(results)

results_df
```
